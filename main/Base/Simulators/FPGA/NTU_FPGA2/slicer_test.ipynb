```python
import random

P = 64

def u128_hex(x):
    return f"0x{x:032x}"

def parse_u128(val):
    if isinstance(val, str):
        s = val.lower().replace("0x", "").replace("128'h", "").replace("_", "")
        return int(s, 16)
    return int(val)

def s64_hex(x):
    return f"0x{(x & ((1 << 64) - 1)):016x}"

def to_signed(val, bits):
    if val & (1 << (bits - 1)):
        return val - (1 << bits)
    return val

def slice_q(a_u128, IA, IB, IOUT):
    FRAC_OUT = P - IOUT
    FRAC_A = P - IA
    FRAC_B = P - IB
    LB = (FRAC_A + FRAC_B) - FRAC_OUT
    HB = LB + P - 1

    mask = (1 << P) - 1
    slice_u = (a_u128 >> LB) & mask
    slice_s = to_signed(slice_u, P)
    return slice_u, slice_s

# random 128-bit input
#a_u128 = random.getrandbits(128)
a_u128 = parse_u128("128'h00389065A3C107FDBF20A653C1900000")
print("a_u128 =", u128_hex(a_u128))

# Q4.60
u_q460, s_q460 = slice_q(a_u128, IA=4, IB=4, IOUT=4)
print("Q4.60  u64 =", s64_hex(u_q460), "signed =", s_q460)

# Q3.61
u_q361, s_q361 = slice_q(a_u128, IA=3, IB=3, IOUT=3)
print("Q3.61  u64 =", s64_hex(u_q361), "signed =", s_q361)
```
